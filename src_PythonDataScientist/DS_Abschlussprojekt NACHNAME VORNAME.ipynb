#### Train-Test-Split

#### Algorithmusauswahl

-   Wähle genau 2 Klassifikationsalgorithmen, die keine Skalierung
    benötigen

    -   Falls der Algorithmus das Argument `random_state` besitzt →

    -   Falls der Algorithmus das Argument `max_iter` besitzt → setze
        auf `10000`

#### Modellbewertung (nur Trainingsdatensatz)

-   Führe auf den Trainingsdatensatz eine Repeated Stratified K-Fold
    Cross-Validation durch:
    -   `n_splits=5`
    -   `n_repeats=3`

-   Bewertungsmetrik: Accuracy
-   Erstelle eine Tabelle mit Min, Mean, Max der Accuracy je Modell
-   Ermittle das Modell mit der höchsten mittleren CV-Accuracy

#### Finaler Holdout-Test

-   Trainiere das soeben ermittelte, beste Modell nochmal, aber nun auf
    dem gesamten Trainingsdatensatz (ohne weitere CV-Unterteilung)
-   Evaluiere dann damit auf dem Holdout-Testdatensatz
    -   die Accuracy
    -   die Confusion Matrix
